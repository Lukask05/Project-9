```python
# import the libraries we need
import spacy
import operator
import urllib.request

# load the spacy english language model
nlp = spacy.load('en_core_web_sm')
```

```python
# this function downloads text from a url
def fetch_text(url):
    """Downloads and returns the text from a given url."""
    with urllib.request.urlopen(url) as response:
        raw_bytes = response.read()
    return raw_bytes.decode('utf-8')


# download the text from Project Gutenberg
url = 'https://www.gutenberg.org/cache/epub/132/pg132.txt'
raw_text = fetch_text(url)

# Project Gutenberg adds extra header and footer text that we don't want
# so we cut the text down to just the actual book content
start_marker = '*** START OF THE PROJECT GUTENBERG EBOOK'
end_marker = '*** END OF THE PROJECT GUTENBERG EBOOK'

start_idx = raw_text.find(start_marker)
end_idx = raw_text.find(end_marker)

# move past the start marker line to get to the actual text
start_idx = raw_text.find('\n', start_idx) + 1
text = raw_text[start_idx:end_idx].strip()

print("Text loaded!")
print("Total characters:", len(text))
print()
print("First 300 characters:")
print(text[:300])
```

```python
# this function cleans up the text and returns a list of words
# it removes punctuation, numbers, and common words like 'the' and 'is'
def word_tokenization_normalization(text):
    """Takes a string of text and returns a cleaned list of words."""
    doc = nlp(text)
    words = []
    for token in doc:
        # only keep real words, skip stop words like 'the', 'a', 'is'
        if token.is_alpha and not token.is_stop:
            words.append(token.text.lower())
    return words


# this function counts how many times each word appears
def word_count(word_list):
    """Takes a list of words and returns a dictionary of word counts."""
    counts = {}
    for word in word_list:
        if word in counts:
            counts[word] = counts[word] + 1
        else:
            counts[word] = 1
    # sort the dictionary so the most common words come first
    sorted_counts = dict(sorted(counts.items(), key=operator.itemgetter(1), reverse=True))
    return sorted_counts


# this function prints the top 15 words in a readable way
def print_top_15_frequent_words(word_counts):
    """Prints the top 15 most frequent words and how many times they appear."""
    rank = 1
    for word, count in list(word_counts.items())[:15]:
        print(str(rank) + ". " + word + " - " + str(count))
        rank = rank + 1
```

```python
# run the word frequency analysis on our text

# increase the max length so spacy can handle the full book
nlp.max_length = 2000000

# clean the text and get a word list
word_list = word_tokenization_normalization(text)

# count each word
word_counts = word_count(word_list)

print("Total words after cleaning:", len(word_list))
print("Unique words:", len(word_counts))
print()
print("=== Top 15 Most Frequent Words in The Art of War ===")
print()
print_top_15_frequent_words(word_counts)
```

## Part 3: Word Frequency Analysis

Looking at the top 15 words from The Art of War, the results make a lot of sense for this type of book. Enemy, general, army, and attack show up near the top of the list, which makes since the whole book is about military strategy and winning battles. I was a little surprised to see heaven and earth in the top 15. After thinking about it though, Sun Tzu talks a lot about using nature and the environment to your advantage, so those words make sense as part of his strategy advice. Also the Chinese were very into mythology so I guess heaven makes sense. The word men also shows up a lot which reflects how the book constantly talks about controlling and directing soldiers as a group. One thing that surprised me is that words like victories or winning did not make the top 15, even though the whole point of the book is about winning. It seems like Sun Tzu focuses more on the process of strategy rather than just talking about winning directly. Overall I would say the top words do a good job of showing what the book is really about.

```python
# Part 4: Extended Analysis - Top 10 Verbs
# spacy can identify what part of speech each word is
# we will use that to find the most common verbs in the text

def get_top_verbs(text, n=10):
    """Finds the top n most common verbs in a text. Returns a list of (verb, count) pairs."""
    doc = nlp(text)
    verb_counts = {}

    for token in doc:
        # check if the word is a verb and not a stop word
        if token.pos_ == 'VERB' and token.is_alpha and not token.is_stop:
            # use the lemma (base form) so 'attacked' and 'attack' count as the same word
            base_verb = token.lemma_.lower()
            if base_verb in verb_counts:
                verb_counts[base_verb] = verb_counts[base_verb] + 1
            else:
                verb_counts[base_verb] = 1

    # sort and return just the top n
    top_verbs = sorted(verb_counts.items(), key=operator.itemgetter(1), reverse=True)[:n]
    return top_verbs


# run the verb analysis
top_verbs = get_top_verbs(text, n=10)

print("=== Top 10 Most Frequent Verbs in The Art of War ===")
print()
rank = 1
for verb, count in top_verbs:
    print(str(rank) + ". " + verb + " - " + str(count))
    rank = rank + 1
```

## Part 4: Extended Analysis - Top 10 Verbs

I chose to find the top 10 verbs because I wanted to see what kinds of actions Sun Tzu talks about most. The results were really interesting action words like attack, advance, and move showed up near the top, which makes sense for a book about warfare. I was also surprised to see know as one of the most common verbs. This fits with the famous idea from the book that knowing yourself and your enemy is the key to winning. The verb list shows that the book is not just about physical fighting, but also about thinking and planning.
